# Exploratory 16 x 16 experiments

This notebook contains the support comparison and the inducing-spacing experiment.


```python
from pathlib import Path
import subprocess
import sys

REPO = Path.cwd()
if REPO.name == "notebooks":
    REPO = REPO.parent

def run(*args):
    subprocess.run([sys.executable, *args], cwd=REPO, check=True)

SEEDS = [0, 1, 2]
```

## Run controls

Both experiments use 200,000 pretraining steps and two NUTS chains with 4,000 warmup iterations and 6,000 retained draws per chain.


```python
RUN_SUPPORT_COMPARISON = True
RUN_SPACING_COMPARISON = False
```

## Exact, low-resolution and local support comparison


```python
if RUN_SUPPORT_COMPARISON:
    for seed in SEEDS:
        run(
            "experiments/exploratory_16x16.py",
            "--seed", str(seed),
            "--mcmc-warmup", "4000",
            "--mcmc-samples", "6000",
            "--num-chains", "2",
        )
```

## Inducing-spacing comparison

The retained lengthscales are 10, 30 and 50; the inducing grid sides are 4, 6, 8, 12 and 16.


```python
if RUN_SPACING_COMPARISON:
    for seed in SEEDS:
        run(
            "experiments/exploratory_16x16_spacing.py",
            "--seed", str(seed),
            "--lengthscales", "10", "30", "50",
            "--pretrain-grid-sizes", "4", "6", "8", "12", "16",
            "--mcmc-warmup", "4000",
            "--mcmc-samples", "6000",
            "--num-chains", "2",
        )
```
