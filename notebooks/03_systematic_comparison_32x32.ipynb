# 32 x 32 systematic comparison

This notebook compares the Full GP reference, Exact DeepRV and full-domain Bilinear, Cubic, DTC and FITC DeepRV models for inducing sides 4, 8 and 16.


```python
from pathlib import Path
import subprocess
import sys

REPO = Path.cwd()
if REPO.name == "notebooks":
    REPO = REPO.parent

def run(*args):
    subprocess.run([sys.executable, *args], cwd=REPO, check=True)

SEEDS = [0, 1, 2]
INDUCING_SIDES = (4, 8, 16)
```

## Run controls

All posterior runs use two chains with 1,000 warmup iterations and 4,000 retained draws per chain.


```python
RUN_DEEPRV_COMPARISON = True
```

## DeepRV teacher comparison and Full GP reference


```python
if RUN_DEEPRV_COMPARISON:
    for seed in SEEDS:
        run(
            "experiments/grid32.py",
            "--grid-size", "32",
            "--inducing-grid-sizes", *map(str, INDUCING_SIDES),
            "--seed", str(seed),
            "--obs-mask-type", "uniform",
            "--train-steps", "200000",
            "--batch-size", "32",
            "--validation-interval", "500",
            "--validation-batches", "500",
            "--checkpoint-save-interval", "10000",
            "--num-chains", "2",
            "--initial-warmup", "1000",
            "--initial-samples", "4000",
            "--target-accept-prob", "0.8",
            "--max-tree-depth", "10",
            "--init-to-median-num-samples", "10",
        )
```
