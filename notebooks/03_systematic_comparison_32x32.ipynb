# 32 x 32 systematic comparison

This notebook groups the DeepRV teacher comparison, Full GP reference and matched Direct GP comparison for inducing sides 4, 8 and 16.


```python
from pathlib import Path
import subprocess
import sys

REPO = Path.cwd()
if REPO.name == "notebooks":
    REPO = REPO.parent

def run(*args):
    subprocess.run([sys.executable, *args], cwd=REPO, check=True)

SEEDS = [0, 1, 2]
INDUCING_SIDES = (4, 8, 16)
```

## Run controls

All posterior runs use two chains with 1,000 warmup iterations and 4,000 retained draws per chain.


```python
RUN_DEEPRV_COMPARISON = True
RUN_DIRECT_GP_COMPARISON = False
```

## DeepRV teacher comparison and Full GP reference


```python
if RUN_DEEPRV_COMPARISON:
    for seed in SEEDS:
        run(
            "experiments/grid32.py",
            "--grid-size", "32",
            "--inducing-grid-sizes", *map(str, INDUCING_SIDES),
            "--seed", str(seed),
            "--obs-mask-type", "uniform",
            "--num-chains", "2",
            "--initial-warmup", "1000",
            "--initial-samples", "4000",
            "--extended-warmup", "1000",
            "--extended-samples", "4000",
        )
```

## Matched Direct GP comparison


```python
if RUN_DIRECT_GP_COMPARISON:
    for seed in SEEDS:
        for side in INDUCING_SIDES:
            run(
                "experiments/grid32_direct_gp.py",
                "--grid-size", "32",
                "--inducing-grid-size", str(side),
                "--seed", str(seed),
                "--obs-mask-type", "uniform",
                "--num-chains", "2",
                "--initial-warmup", "1000",
                "--initial-samples", "4000",
                "--extended-warmup", "1000",
                "--extended-samples", "4000",
            )
```
