# Target-128 experiments

This notebook groups Exact128, the Grid-64 DeepRV models and the Full GP128 feasibility probe. The Full GP section is a runtime probe only and is not a formal posterior experiment.


```python
from pathlib import Path
import subprocess
import sys

REPO = Path.cwd()
if REPO.name == "notebooks":
    REPO = REPO.parent

def run(*args):
    subprocess.run([sys.executable, *args], cwd=REPO, check=True)

SEEDS = [0, 1, 2]
```

## Run controls

Target-128 posterior inference uses one chain with 4,000 warmup iterations and 6,000 retained draws. All stages are disabled by default because they require substantial GPU memory and runtime.


```python
RUN_EXACT_TRAINING = False
RUN_EXACT_INFERENCE = False
RUN_GRID64_TRAINING = False
RUN_GRID64_INFERENCE = False
RUN_FULL_GP_RUNTIME_PROBE = False
```

## Exact128


```python
if RUN_EXACT_TRAINING:
    run("experiments/target128_exact.py", "--stage", "train")

if RUN_EXACT_INFERENCE:
    run("experiments/target128_exact.py", "--stage", "infer")
```

## Bilinear64, Cubic64, DTC64 and FITC64


```python
if RUN_GRID64_TRAINING:
    run("experiments/target128_grid64.py", "--stage", "train")

if RUN_GRID64_INFERENCE:
    run("experiments/target128_grid64.py", "--stage", "infer")
```

## Full GP128 feasibility and runtime probe

The short chains below measure feasibility only. Their samples must not be interpreted as a formal posterior result.


```python
if RUN_FULL_GP_RUNTIME_PROBE:
    for seed in SEEDS:
        run(
            "experiments/full_gp128_probe.py",
            "--seed", str(seed),
            "--warmup", "10",
            "--samples", "10",
        )
```
