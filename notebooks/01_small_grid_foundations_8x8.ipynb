# Exploratory 8 x 8 experiments

This notebook groups the exploratory model comparison and the MCMC budget comparison. The scientific implementations remain in the corresponding experiment scripts.


```python
from pathlib import Path
import subprocess
import sys

REPO = Path.cwd()
if REPO.name == "notebooks":
    REPO = REPO.parent

def run(*args):
    subprocess.run([sys.executable, *args], cwd=REPO, check=True)

SEEDS = [0, 1, 2]
```

## Run controls

The main comparison uses two chains with 4,000 warmup iterations and 6,000 retained draws per chain. The optional budget comparison evaluates the short and long schedules explicitly.


```python
RUN_MAIN_COMPARISON = True
RUN_BUDGET_COMPARISON = False
BUDGETS = ("short", "long")
```

## Full GP, Exact DeepRV, low-resolution and local comparison


```python
if RUN_MAIN_COMPARISON:
    for seed in SEEDS:
        run(
            "experiments/exploratory_8x8.py",
            "--seed", str(seed),
            "--mcmc-warmup", "4000",
            "--mcmc-samples", "6000",
            "--num-chains", "2",
        )
```

## MCMC budget comparison

Short: 200 warmup and 1,000 retained draws per chain. Long: 4,000 warmup and 6,000 retained draws per chain.


```python
if RUN_BUDGET_COMPARISON:
    for seed in SEEDS:
        run(
            "experiments/exploratory_8x8_mcmc.py",
            "--seed", str(seed),
            "--budgets", *BUDGETS,
            "--num-chains", "2",
            "--short-warmup", "200",
            "--short-samples", "1000",
            "--long-warmup", "4000",
            "--long-samples", "6000",
        )
```
