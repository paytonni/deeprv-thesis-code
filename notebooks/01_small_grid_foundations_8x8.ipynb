# Exploratory 8 x 8 experiments

This notebook runs the thesis exploratory comparison between Full GP, Exact DeepRV, low-resolution DeepRV and local-support DeepRV.


```python
from pathlib import Path
import subprocess
import sys

REPO = Path.cwd()
if REPO.name == "notebooks":
    REPO = REPO.parent

def run(*args):
    subprocess.run([sys.executable, *args], cwd=REPO, check=True)

SEEDS = [0, 1, 2]
```

## Run controls

The comparison uses two chains with 4,000 warmup iterations and 6,000 retained draws per chain.


```python
RUN_MAIN_COMPARISON = True
```

## Full GP, Exact DeepRV, low-resolution and local comparison


```python
if RUN_MAIN_COMPARISON:
    for seed in SEEDS:
        run(
            "experiments/exploratory_8x8.py",
            "--seed", str(seed),
            "--mcmc-warmup", "4000",
            "--mcmc-samples", "6000",
            "--num-chains", "2",
        )
```
