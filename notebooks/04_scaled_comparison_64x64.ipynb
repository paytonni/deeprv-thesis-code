# 64 x 64 scaled comparison

This notebook groups all three datasets, the DeepRV teacher comparison, Full GP reference and matched Direct GP comparison for inducing sides 8, 16 and 32.


```python
from pathlib import Path
import subprocess
import sys

REPO = Path.cwd()
if REPO.name == "notebooks":
    REPO = REPO.parent

def run(*args):
    subprocess.run([sys.executable, *args], cwd=REPO, check=True)

SEEDS = [0, 1, 2]
INDUCING_SIDES = (8, 16, 32)
```

## Run controls

All posterior runs use two chains with 1,000 warmup iterations and 4,000 retained draws per chain. Set `SEEDS_TO_RUN` to a subset when distributing long runs across machines.


```python
SEEDS_TO_RUN = SEEDS
RUN_DEEPRV_COMPARISON = True
RUN_DIRECT_GP_COMPARISON = False
```

## DeepRV teacher comparison and Full GP reference


```python
if RUN_DEEPRV_COMPARISON:
    for seed in SEEDS_TO_RUN:
        run(
            "experiments/grid64.py",
            "--grid-size", "64",
            "--inducing-grid-sizes", *map(str, INDUCING_SIDES),
            "--seed", str(seed),
            "--obs-mask-type", "uniform",
            "--num-chains", "2",
            "--initial-warmup", "1000",
            "--initial-samples", "4000",
            "--extended-warmup", "1000",
            "--extended-samples", "4000",
        )
```

## Matched Direct GP comparison


```python
if RUN_DIRECT_GP_COMPARISON:
    for seed in SEEDS_TO_RUN:
        for side in INDUCING_SIDES:
            run(
                "experiments/grid64_direct_gp.py",
                "--grid-size", "64",
                "--inducing-grid-size", str(side),
                "--seed", str(seed),
                "--obs-mask-type", "uniform",
                "--num-chains", "2",
                "--initial-warmup", "1000",
                "--initial-samples", "4000",
                "--extended-warmup", "1000",
                "--extended-samples", "4000",
            )
```
